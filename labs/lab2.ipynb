#### 2. Data cleaning

```python
# Missing values and duplicates
print(df.isna().sum())
print('duplicates:', df.duplicated().sum())
```

```python
# Harmonize the categories: same label written in different cases
df['staff_experience'] = df['staff_experience'].str.lower()
df['waste_category'] = df['waste_category'].str.lower()
print(df['staff_experience'].value_counts(dropna=False))
print(df['waste_category'].value_counts())
```

```python
# Temperature: exactly -10 and 60 repeated many times -> sensor placeholder values, not real readings
print(df['temperature_C'].round(2).value_counts().head(3))

# Replace by NaN, then impute with the median:
# - 21 rows only: imputing keeps their other (valid) columns instead of dropping them
# - median rather than mean: not pulled by the remaining extreme values
df['temperature_C'] = df['temperature_C'].mask(df['temperature_C'].isin([-10, 60]))
df['temperature_C'] = df['temperature_C'].fillna(df['temperature_C'].median())
```

```python
# staff_experience: 164 missing values, impute with the most frequent level (mode):
# - ~18% of the rows: dropping them would lose too much data
# - categorical column: mean/median impossible, the mode ("intermediate") is the standard choice
df['staff_experience'] = df['staff_experience'].fillna(df['staff_experience'].mode()[0])
```

```python
# meals_served: describe() shows 75% of days < 407 meals but a max of 4730.
big = df['meals_served'] > 1000
print(big.sum(), 'days > 1000 meals, max of the other days:', df.loc[~big, 'meals_served'].max())
print(df.loc[big, ['meals_served', 'food_waste_kg']].describe())
print('corr meals_served / target:', df['meals_served'].corr(df['food_waste_kg']).round(2))
```

These large values come with a large `food_waste_kg` (strong correlation 0.85): they are consistent real events (big days) so we keep them.

```python
# Check: does the month of the date have a link with the target?
print('corr month / target:', pd.to_datetime(df['date']).dt.month.corr(df['food_waste_kg']).round(2))

# Drop ID: row identifier, no link with food waste
# Drop date: text not usable by a model; the day of the week is already in day_of_week
# and the month has no link with the target (corr ~0), so nothing useful to extract
df = df.drop(columns=['ID', 'date'])
df.head()
```

#### 3. Scaling & transformation

```python
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder

# Ordinal encoding: experience has a natural order
ord_enc = OrdinalEncoder(categories=[['beginner', 'intermediate', 'expert']])
df['staff_experience'] = ord_enc.fit_transform(df[['staff_experience']])

# One-hot encoding: no order between waste categories or between days
oh_enc = OneHotEncoder(sparse_output=False)
oh_cols = ['waste_category', 'day_of_week']
oh = pd.DataFrame(oh_enc.fit_transform(df[oh_cols]),
                  columns=oh_enc.get_feature_names_out(oh_cols), index=df.index)
df = pd.concat([df.drop(columns=oh_cols), oh], axis=1)
```

```python
# Z-score (StandardScaler) for the continuous features, rather than MinMax:
# with MinMax, the kept extreme values of meals_served (max 4730) would squeeze all normal days into [0, 0.09]
std_cols = ['meals_served', 'kitchen_staff', 'temperature_C', 'humidity_percent', 'past_waste_kg', 'staff_experience']
df[std_cols] = StandardScaler().fit_transform(df[std_cols])

# Not scaled: special_event and one-hot columns (already 0/1), target food_waste_kg (kept in kg)
df.head()
```

```python
# Final check: no missing values, all columns numerical
print(df.shape)
print(df.isna().sum().sum(), 'missing values')
print(df.dtypes.value_counts())
```
